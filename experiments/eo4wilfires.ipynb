```python
import os
import tarfile

def main():
    tar_path = r"C:\Users\Vardhan\Desktop\Coding\Minor Project\eo4wildfires_local\eo4wildfires.tar.gz"
    extract_dir = "./eo4wildfires_local/extracted"

    if not os.path.exists(tar_path):
        print(f"Archive file not found at {tar_path}. Ensure download is complete.")
        return

    print("Extracting eo4wildfires.tar.gz (this may take a few minutes)...")
    os.makedirs(extract_dir, exist_ok=True)
    
    with tarfile.open(tar_path, "r:gz") as tar:
        tar.extractall(path=extract_dir)
        
    print(f"Successfully extracted dataset to '{extract_dir}'!")

if __name__ == "__main__":
    main()
```
